```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

np.random.seed(42)
n = 300
hora = np.random.randint(0, 24, n)
monto = np.random.exponential(scale=150, size=n) + 10

# Fraude más probable de madrugada (0-5h, 23h) y en montos altos
prob_fraude = 0.03 + 0.35 * ((hora <= 5) | (hora >= 23)) + 0.25 * (monto > 400)
prob_fraude = np.clip(prob_fraude, 0, 0.9)
es_fraude = (np.random.rand(n) < prob_fraude).astype(int)

transacciones = pd.DataFrame({"hora": hora, "monto": monto, "es_fraude": es_fraude})
colores = {0: "#0ca30c", 1: "#d03b3b"}
etiquetas = {0: "Legítima", 1: "Fraude"}

fig, ax = plt.subplots(figsize=(10, 6), facecolor="#fcfcfb")
ax.set_facecolor("#fcfcfb")

for valor in [0, 1]:
    subset = transacciones[transacciones["es_fraude"] == valor]
    ax.scatter(
        subset["hora"], subset["monto"],
        s=70, alpha=0.75,
        c=colores[valor],
        edgecolors="white", linewidths=0.6,
        label=etiquetas[valor]
    )

ax.set_title("Transacciones por hora y monto", fontsize=15, color="#0b0b0b", pad=14)
ax.set_xlabel("Hora del día", fontsize=11, color="#52514e")
ax.set_ylabel("Monto (S/)", fontsize=11, color="#52514e")
ax.set_xticks(range(0, 24, 2))
ax.grid(True, color="#e1e0d9", linewidth=0.8)
ax.spines[["top", "right"]].set_visible(False)
ax.legend(title="Tipo de transacción", frameon=False, loc="upper right")

plt.tight_layout()
plt.show()

print("Total transacciones:", len(transacciones), "Fraudes:", transacciones["es_fraude"].sum())
```

### 5.2. Ficha PEAS del Termostato Inteligente
* **Percepción (S):** Temperatura actual medida en grados Celsius (°C).
* **Acciones (A):** Encender aire acondicionado, encender calefacción, apagar climatización.
* **Entorno (E):** Habitación cerrada / ambiente interior.
* **Objetivo:** Mantener la temperatura en el rango de confort deseado (22°C ± 1°C).
* **Medida de desempeño (P):** Porcentaje de tiempo en temperatura de confort minimizando el consumo energético.

### 5.3. Reglas de Decisión
* Si la temperatura está muy por encima de la deseada (> 23.0°C) -> Encender aire acondicionado.
* Si la temperatura está muy por debajo de la deseada (< 21.0°C) -> Encender calefacción.
* Si la temperatura está cerca de la deseada (entre 21.0°C y 23.0°C) -> Apagar climatización.

```python
def agente_termostato(temperatura_actual, temperatura_objetivo=22.0, margen=1.0):
    """
    Agente reactivo simple para control de climatización.
    """
    diferencia = temperatura_actual - temperatura_objetivo

    if diferencia > margen:
        return "Encender aire acondicionado"
    elif diferencia < -margen:
        return "Encender calefaccion"
    else:
        return "Apagar climatizacion"
```

```python
percepciones = [18.0, 19.5, 21.8, 24.3, 26.0, 23.0, 20.5, 21.0, 28.5]

for t in percepciones:
    accion = agente_termostato(t)
    print(f"Percepcion: {t}°C -> Accion del agente: {accion}")
```

### 5.6. Reflexión
* **¿Es racional?** Sí, porque bajo su conjunto de percepciones inmediatas siempre ejecuta la acción que maximiza su medida de desempeño esperada.
* **¿Bajo qué condiciones dejaría de serlo?**
  1. Si el sensor se descalibra o genera ruido actuando sobre lecturas erredas.
  2. Si el margen es inadecuado: muy estrecho provocará encendidos y apagados continuos, y muy ancho perderá la condición de confort.
  3. Si existe una perturbación constante, ya que al carecer de memoria mantendrá la calefacción encendida sin detectar la anomalía.
