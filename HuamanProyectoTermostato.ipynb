```python
def agente_termostato(temperatura_actual, temperatura_objetivo=22, margen=1.0):
    """
    Percepcion(S): temperatura que sera medida por el sensor
    Acciones(A): Encender aire acondicionado, encender calefacción, apagar climatización.
    Entorno (E): Habitación cerrada / ambiente interior.
    Medida de desempeño (P): Ahorro de energia, tiempo que puede optimizar la temperatura.
    """
    diferencia = temperatura_actual - temperatura_objetivo
    if diferencia > margen:
        return "Encender aire acondicionado"
    elif diferencia < -margen:
        return "Encender calefaccion"
    else:
        return "Apagar climatizacion"
```

```python
percepciones = [18.0, 19.5, 21.8, 24.3, 26.0, 23.0, 20.5]
for t in percepciones:
    accion = agente_termostato(t)
    print(f"Percepcion: {t} C -> Accion del agente: {accion}")
```

## Ficha PEAS: agente de riego automático

| Elemento | Descripción |
|---|---|
| Percepción (S) | Porcentaje de humedad del suelo y pronóstico de lluvia para el día. |
| Acciones (A) | Regar, no regar o emitir una alerta de revisión. |
| Entorno (E) | Jardín, plantas, suelo, sensor de humedad, sistema de riego y servicio de pronóstico meteorológico. |
| Objetivo | Mantener una humedad adecuada para las plantas evitando desperdiciar agua. |
| Medida de desempeño (P) | Salud de las plantas, ahorro de agua, humedad apropiada y reducción del riego innecesario. |

```python
def agente_riego(humedad_suelo, llovera_hoy):
    """
    Agente reactivo simple para decidir el riego de un jardín.

    Parámetros:
    - humedad_suelo: número entre 0 y 100.
    - llovera_hoy: True si se pronostica lluvia; False en caso contrario.
    """

    if humedad_suelo < 30 and not llovera_hoy:
        return "Regar el jardín"
    elif humedad_suelo < 30 and llovera_hoy:
        return "No regar: el suelo está seco, pero se espera lluvia"
    elif humedad_suelo > 85:
        return "No regar y revisar drenaje: exceso de humedad"
    else:
        return "No regar: humedad adecuada"
```

```python
pruebas_riego = [
    (20, False),
    (20, True),
    (45, False),
    (70, True),
    (90, False),
    (30, False)
]

for humedad, lluvia in pruebas_riego:
    accion = agente_riego(humedad, lluvia)
    print(
        f"Humedad: {humedad}% | ¿Lloverá hoy?: {lluvia} "
        f"-> Acción: {accion}"
    )
```

## Reflexión sobre el agente de riego

El diseño del agente de riego fue más complejo que el del termostato porque considera dos percepciones simultáneamente: humedad del suelo y pronóstico de lluvia. Por ello, las reglas deben combinar condiciones mediante operadores lógicos como `and` y `not`.

Sigue siendo un agente reactivo simple porque toma decisiones basándose únicamente en las percepciones actuales. No conserva información histórica sobre días anteriores, cantidad de agua usada, tipo de planta ni evolución de la humedad.
